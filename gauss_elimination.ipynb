```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def gauss_elimination(A,B):
    augmented=np.column_stack((A,B))
    n=len(augmented)
    for i in range(n):
        pivot_row=i
        for k in range(i+1,n):
            if abs(augmented[k][i]) > abs(augmented[pivot_row][i]):
                pivot_row=k
        temp = augmented[i].copy()
        
        augmented[i] = augmented[pivot_row]
        
        augmented[pivot_row] = temp
        
        pivot=augmented[i][i]
        
        for j in range(i+1, n):
            
            facteur=augmented[j][i]/pivot
            
            augmented[j]=augmented[j]-(facteur)*augmented[i]
            
    proof=augmented.copy()
    
    x = np.zeros(n)

    x[n-1] = augmented[n-1][n] / augmented[n-1][n-1]

    for i in range(n-2, -1, -1):
        somme = 0

        for j in range(i+1, n):
            somme += augmented[i][j] * x[j]

        x[i] = (augmented[i][n] - somme) / augmented[i][i]
    
    return x,proof    
```

```python
A = np.array([
    [2,  1, -1],
    [-3, -1,  2],
    [-2,  1,  2]
], dtype=float)

B = np.array([8, -11, -3], dtype=float)

resultat,proof = gauss_elimination(A, B)

print(proof)
print(resultat)
```
